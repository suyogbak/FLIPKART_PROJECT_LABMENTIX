## **DATA DICTIONARY**

1. **`Unique id`** – Unique identifier assigned to each customer interaction record
   Helps distinguish individual customer service cases and ensures record-level traceability.

2. **`Channel name`** – Customer service channel through which the interaction was initiated
   Represents the communication medium such as  email, inbound or outbound support channels.

3. **`Category`** – High-level classification of the customer issue
   Groups customer interactions into broader issue categories for operational analysis.

4. **`Sub-category`** – Detailed classification of the customer issue
   Provides more granular information about the specific nature of the customer's concern.

5. **`Customer Remarks`** – Feedback or comments provided by the customer
   Contains valuable qualitative insights regarding customer experience, concerns, and satisfaction.

6. **`Order id`** – Unique identifier associated with the customer's order
   Links the customer interaction to a specific order within the Flipkart ecosystem.

7. **`Order date time`** – Date and time when the order was placed
   Helps analyze customer support requests relative to the order lifecycle.

8. **`Issue reported at`** – Timestamp when the customer reported the issue
   Indicates when the support request was initiated by the customer.

9. **`Issue responded`** – Timestamp when the customer issue received a response
   Used to evaluate response time and support efficiency.

10. **`Survey response date`** – Date on which the customer submitted the satisfaction survey
    Captures when the customer provided feedback regarding the support experience.

11. **`Customer city`** – City of the customer
    Enables geographical analysis of customer satisfaction and support performance.


12. **`Product category`** – Category of the purchased product
    Helps identify product segments associated with higher or lower customer satisfaction levels.

13. **`Item price`** – Price of the purchased item
    Represents the monetary value of the order and may influence customer expectations and satisfaction.

14. **`Connected handling time`** – Total time spent handling the customer interaction
    Measures the duration required by support agents to address customer concerns.

15. **`Agent name`** – Name of the customer service representative handling the interaction
    Enables performance analysis at the individual agent level.

16. **`Supervisor`** – Name of the supervisor overseeing the support agent
    Helps evaluate team-level performance and management effectiveness.

17. **`Manager`** – Name of the manager responsible for the support team
    Facilitates higher-level operational and performance analysis.

18. **`Tenure Bucket`** – Categorized experience level of the support agent
    Indicates the duration of service of the agent and helps assess the impact of experience on customer satisfaction.

19. **`Agent Shift`** – Working shift assigned to the support agent
    Allows analysis of customer satisfaction across different operational shifts.

20. **`CSAT Score`** – Customer Satisfaction Score ranging from 1 to 5
    Represents the target variable of the project and reflects the customer's overall satisfaction with the support experience.

```python
# Get the value counts of 'channel_name'
channel_counts = data['channel_name'].value_counts()

# Create a figure and a set of subplots with a specified figure size
fig, ax = plt.subplots(figsize=(7, 7)) # Adjusted figure size for a slightly smaller pie
# Generate a pie chart from the value counts, add labels, display percentages, set the starting angle, and apply a color palette
ax.pie(channel_counts, labels=channel_counts.index, autopct='%1.2f%%', startangle=90, colors=sns.color_palette('tab10'))
# Set the title of the pie chart with a specified font size
ax.set_title('Distribution of Customer Query Channels', fontsize=18) # Slightly reduced title font size
# Ensure the pie chart is drawn as a circle (equal aspect ratio)
ax.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.

# Add a background color to the figure for visual consistency
fig.patch.set_facecolor('#f0f0f0') # Light grey background

# Display the generated plot
plt.show()
```

```python
# List of columns from the DataFrame to iterate through and visualize their distributions.
cols_to_plot = ['category', 'Tenure Bucket', 'Manager', 'Agent Shift', 'CSAT Score']

# Loop through each column specified in the 'cols_to_plot' list.
for col in cols_to_plot:
    # Calculate the frequency of each unique value in the current column.
    counts = data[col].value_counts()

    # Check if the number of unique values in the current column is greater than 6.
    if len(counts) > 6:
        # If it is, take the top 5 most frequent values.
        top_5 = counts[:5]
        # Group the remaining less frequent values into an 'Others' category.
        others = pd.Series({'Others': counts[5:].sum()})
        # Concatenate the top 5 values with the 'Others' category for plotting.
        plot_data = pd.concat([top_5, others])
    else:
        # If there are 6 or fewer unique values, use all of them directly for plotting.
        plot_data = counts

    # Create a new figure and a single subplot (axis) for the pie chart.
    fig, ax = plt.subplots(figsize=(5, 5))
    # Generate a pie chart using the 'plot_data'.
    # 'labels' are the categories, 'autopct' formats percentages on slices,
    # 'startangle' rotates the start of the first slice, and 'colors' sets the color palette.
    ax.pie(plot_data, labels=plot_data.index, autopct='%1.2f%%', startangle=90, colors=sns.color_palette('tab10'))
    # Set the title of the pie chart dynamically based on the current column name.
    ax.set_title(f'Distribution of {col}', fontsize=18)
    # Ensure the pie chart is drawn as a perfect circle.
    ax.axis('equal')

    # Set the background color of the entire figure for aesthetic consistency.
    fig.patch.set_facecolor('#f0f0f0')

    # Adjust plot parameters for a tight layout, preventing labels from overlapping.
    plt.tight_layout()
    # Display the generated plot.
    plt.show()
```

```python
# Sub-categorizing 'Returns' and 'Order Related' queries

target_categories = ['Returns', 'Order Related']

for cat in target_categories:
    # Filter data for the specific category
    subset_data = data[data['category'] == cat]
    counts = subset_data['Sub-category'].value_counts()

    # Logic for 'Others' if unique sub-categories > 6
    if len(counts) > 6:
        top_5 = counts[:5]
        others = pd.Series({'Others': counts[5:].sum()})
        plot_data = pd.concat([top_5, others])
    else:
        plot_data = counts

    # Create visualization
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.pie(plot_data, labels=plot_data.index, autopct='%1.2f%%', startangle=90, colors=sns.color_palette('tab10'))
    ax.set_title(f'Sub-category Distribution for: {cat}', fontsize=18)
    ax.axis('equal')

    # Setting background color
    fig.patch.set_facecolor('#f0f0f0')

    plt.tight_layout()
    plt.show()
```
